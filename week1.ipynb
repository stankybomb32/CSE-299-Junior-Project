```python
from ucimlrepo import fetch_ucirepo

# fetch dataset
breast_cancer_wisconsin_diagnostic = fetch_ucirepo(id=17)

# data (as pandas dataframes)
X = breast_cancer_wisconsin_diagnostic.data.features
y = breast_cancer_wisconsin_diagnostic.data.targets

# metadata
print(breast_cancer_wisconsin_diagnostic.metadata)

# variable information
print(breast_cancer_wisconsin_diagnostic.variables)
```

# Preprocessing

```python
X = X[['perimeter3','area3','concavity3']] #Selecting a set number of features
X = (X - X.mean()) / X.std() # Standardizing so each feature is on a similar scale
y = y['Diagnosis'].map({'M': 1, 'B': 0}) # Assigning Binary Values

X.head()
```

# Importing NumPy

```python
import numpy as np
```

## Defining Functions

```python
# initial weights, one per input feature
np.random.seed(42)
weights = np.random.randn(1, 3)   # or np.random.rand(1, 3) for [0, 1)

print(weights)
# perimeter3 area3 concavity3
```

## Defining Activation Functions

```python
def relu(x):
    return np.maximum(0, x)


def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

## Defining Neural Network

```python
alpha = 0.1 # determines step size
def neural_network(input, weights):
  pred = input
  for i in weights:
    pred = pred.dot(i)
  pred = sigmoid(pred) #hard coding one activation layer
  return pred
```

## Testing


## Predict

```python
y_pred = neural_network(X.iloc[0], weights)
```

```python
error = (y_pred - y.iloc[0]) ** 2
delta = (y_pred - y.iloc[0])
```

```python
print(error)
```

## Compare and Learn

```python
X_np = X.values.astype(float)
y_np = y.values.astype(float).ravel()

alpha = 0.01

for epoch in range(100):
  total_error = 0
  for i in range(len(X_np)):
    input = X_np[i]
    goal = y_np[i]

    pred = neural_network(input, weights)
    error = (pred - goal) ** 2
    delta = pred - goal

    weight_delta = input * delta
    weights = weights - alpha * weight_delta

    total_error += error

  if epoch % 10 == 0:
    print(f"Epoch {epoch}: mean error = {total_error / len(X_np):.4f}")

print("Final weights:", weights)
```

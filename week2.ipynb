# Preprocessing

```python
X = X[['perimeter3','area3','concavity3']] #Selecting a set number of features
X = (X - X.mean()) / X.std() # Standardizing so each feature is on a similar scale
y = y['Diagnosis'].map({'M': 1, 'B': 0}) # Assigning Binary Values

X.head()
```

# Importing NumPy

```python
import numpy as np
```

## Defining Functions

```python
hidden_size = 4
alpha = 0.01
```

```python
# initial weights, one per input feature
np.random.seed(42)
weights_01 = np.random.random((X.shape[1], hidden_size))
weights_12 = np.random.random((hidden_size, 1))
weights = [weights_01, weights_12]

print(weights)
# perimeter3 area3 concavity3
```

## Defining Functions

```python
def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu2deriv(output):
  return output > 0

def layer(input, weights):
  return input.dot(weights)
```

## Defining Neural Network

```python
def neural_network(input, weights):
  hidden = relu(layer(input, weights[0]))
  pred = layer(hidden, weights[1])
  return hidden, pred
```

## Compare and Learn

```python
X_np = X.values.astype(float)
y_np = y.values.astype(float).ravel()

alpha = 0.01

for epoch in range(100):
  total_error = 0
  for i in range(len(X_np)):
    input = X_np[i]                               # (3,)
    goal = y_np[i]

    # forward pass
    hidden, pred = neural_network(input, weights)

    error = ((pred - goal) ** 2).sum()

    # backward pass
    delta_2 = pred - goal                                       # (1,)
    delta_1 = delta_2.dot(weights[1].T) * relu2deriv(hidden)    # (hidden_size,)

    weights[1] = weights[1] - alpha * np.outer(hidden, delta_2) # (hidden_size, 1)
    weights[0] = weights[0] - alpha * np.outer(input, delta_1)  # (3, hidden_size)

    total_error += error
  if epoch % 10 == 0:
    print(f"Epoch {epoch}: mean error = {total_error / len(X_np):.4f}")

print("Final weights:", weights)
```
